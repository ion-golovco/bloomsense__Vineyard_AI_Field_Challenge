# Vineyard lab

Outline plots over the Sireț3 imagery, review model output, then re-run and re-judge.

- **Outline plots.** Pick a class in *Draw marks*, then draw a polygon with the toolbar (click each corner, click the first point to close). Classes: `vineyard` (bare-soil inter-rows, the target), `overgrown` (vineyard with grassy inter-rows), `orchard`, `other` (field, garden, grass). Draw two plots when rows change direction, even without a road between them.
- **Fix a mark.** Click it, then *Delete selected mark*; *Undo last* removes your latest one.
- **Layers** (top left) hide or show the model's magenta plots, the organizer layers and the reference tiles.
- **Model output.** Click a magenta plot, then *Right* or *Wrong*; click anywhere to pick that tile for *Vineyard tile* or *No vineyard*. The *missed:* marks record what the model missed.
- Everything goes to `data/review/verdicts.json`. It is evaluation evidence only: it never enters the predictions or the Marcaj upload, because Sireț3 is annotated only in Marcaj.

Start the lab from the repository root with `uv --directory backend run --group lab jupyter lab --ServerApp.root_dir=.. --IdentityProvider.token= --ServerApp.password=` (no password), then open `research/lab.ipynb`.

```python
from marcaj.lab import Lab

lab = Lab(reviewer="")
lab.show()
```

## Re-run the vineyard detector

The window scores are cached, so a re-run takes well under a second and the map updates in place. Parameters: `vine_min`, `vine_over_orchard`, `closing`, `opening` (1 removes strips narrower than one 6.4 m cell), `min_area_m2`, `simplify_m`.

```python
lab.run(vine_min=60, vine_over_orchard=2.0, closing=1, opening=1)
```

## Why were the misses missed?

The detector's scores inside every `block` miss you drew. A low vine median means the row signal is weak there, so it needs a better feature or model, not a lower threshold.
